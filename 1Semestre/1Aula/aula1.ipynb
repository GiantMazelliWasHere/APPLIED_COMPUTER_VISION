```python
# importa a biblioteca
import cv2
```

```python
# define o local da biblioteca
img = cv2.imread("img/imagem_analisar.png")
```

```python
# define o local do retangulo do ROI
cv2.rectangle(img, (720,453),(839,584),(255,0,0),4)
```

```python
# salva a img dps de tratar
cv2.imwrite("img/img_tratada.png",img)
```

```python
# mostra a img tratada
cv2.imshow("imagem_tratada",img)
cv2.waitKey(0)
```

```python
cv2.putText(img,"Busao",(718,437),cv2.FONT_HERSHEY_COMPLEX,1,(255,0,0),1)
```

```python
cv2.imwrite("img/img_processada.png",img)
```

```python
cv2.imshow("imagem_processada",img)
cv2.waitKey(0)
```
